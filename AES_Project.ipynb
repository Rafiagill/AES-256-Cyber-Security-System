```python
import os
import time
from cryptography.hazmat.primitives.ciphers import Cipher, algorithms, modes

# Generate AES key
key = os.urandom(16)

# Read sample file
with open("test_files/sample.txt", "rb") as f:
    data = f.read()

# Generate IV
iv = os.urandom(16)

# Add padding
padding = 16 - len(data) % 16
data += bytes([padding]) * padding

# Create AES cipher
cipher = Cipher(algorithms.AES(key), modes.CBC(iv))
encryptor = cipher.encryptor()

# Start timer
start = time.time()

# Encrypt data
encrypted = encryptor.update(data) + encryptor.finalize()

# End timer
end = time.time()

# Print results
print("✅ Encryption Successful!")
print("⏱ Time Taken:", end - start)
print("🔐 Encrypted Data:")
print(encrypted)
```

```python
import os
import time
from cryptography.hazmat.primitives.ciphers import Cipher, algorithms, modes

# Read sample file
with open("test_files/sample.txt", "rb") as f:
    data = f.read()

# Padding
padding = 16 - len(data) % 16
data += bytes([padding]) * padding

# Number of test runs
runs = 100

total_time = 0

for i in range(runs):

    # Generate key and IV
    key = os.urandom(16)
    iv = os.urandom(16)

    # Create cipher
    cipher = Cipher(algorithms.AES(key), modes.CBC(iv))
    encryptor = cipher.encryptor()

    # Start timer
    start = time.perf_counter()

    # Encrypt data
    encrypted = encryptor.update(data) + encryptor.finalize()

    # End timer
    end = time.perf_counter()

    total_time += (end - start)

# Average execution time
average_time = total_time / runs

print("✅ Optimized AES Performance Test")
print("📁 File Size:", len(data), "bytes")
print("🔁 Total Runs:", runs)
print("⏱ Average Encryption Time:", average_time, "seconds")
```

```python
import os
import time
import pandas as pd
from cryptography.hazmat.primitives.ciphers import Cipher, algorithms, modes

# List of files
files = [
    "test_files/sample.txt",
    "test_files/medium.txt",
    "test_files/large.txt"
]

results = []

for file in files:

    # Read file
    with open(file, "rb") as f:
        data = f.read()

    original_size = len(data)

    # Padding
    padding = 16 - len(data) % 16
    data += bytes([padding]) * padding

    # AES setup
    key = os.urandom(16)
    iv = os.urandom(16)

    cipher = Cipher(algorithms.AES(key), modes.CBC(iv))
    encryptor = cipher.encryptor()

    # Measure time
    start = time.perf_counter()

    encrypted = encryptor.update(data) + encryptor.finalize()

    end = time.perf_counter()

    execution_time = end - start

    # Store results
    results.append({
        "File": os.path.basename(file),
        "File Size (bytes)": original_size,
        "Encryption Time (seconds)": execution_time
    })

# Create table
df = pd.DataFrame(results)

# Display results
print(df)

# Save CSV
df.to_csv("multiple_file_results.csv", index=False)

print("\n✅ Multiple file testing completed!")
```

```python
import hashlib

# Function to calculate SHA-256 hash
def calculate_hash(filename):

    sha256 = hashlib.sha256()

    with open(filename, "rb") as f:

        while True:

            data = f.read(4096)

            if not data:
                break

            sha256.update(data)

    return sha256.hexdigest()

# Calculate original file hash
original_hash = calculate_hash("test_files/sample.txt")

print("✅ SHA-256 Integrity Check")
print("🔐 Original File Hash:\n")
print(original_hash)
```

```python
import hashlib

# Function to generate SHA-256 hash
def calculate_hash(filename):

    sha256 = hashlib.sha256()

    with open(filename, "rb") as f:

        while True:

            data = f.read(4096)

            if not data:
                break

            sha256.update(data)

    return sha256.hexdigest()

# Original hash
original_hash = calculate_hash("test_files/sample.txt")

# Verification hash
verification_hash = calculate_hash("test_files/sample.txt")

print("🔐 Original Hash:")
print(original_hash)

print("\n🔍 Verification Hash:")
print(verification_hash)

# Compare hashes
if original_hash == verification_hash:

    print("\n✅ File Integrity Verified")
    print("🛡 File has NOT been modified")

else:

    print("\n❌ File Tampering Detected")
```

```python
import os
import time
import hashlib
import pandas as pd
import matplotlib.pyplot as plt
from cryptography.hazmat.primitives.ciphers import Cipher, algorithms, modes

# Password
password = "mypassword123"

# Generate AES-256 key
key = hashlib.sha256(password.encode()).digest()

# Test files
files = [
    "test_files/small.txt",
    "test_files/medium.txt",
    "test_files/large.txt"
]

results = []

# Test each file
for file in files:

    # Read file
    with open(file, "rb") as f:
        data = f.read()

    # Generate IV
    iv = os.urandom(16)

    # Add padding
    padding = 16 - len(data) % 16
    data += bytes([padding]) * padding

    # AES Cipher
    cipher = Cipher(algorithms.AES(key), modes.CBC(iv))
    encryptor = cipher.encryptor()

    # Start timer
    start = time.perf_counter()

    # Encrypt
    encrypted = encryptor.update(data) + encryptor.finalize()

    # End timer
    end = time.perf_counter()

    encryption_time = end - start

    # File size
    file_size = len(data)

    # Save results
    results.append([os.path.basename(file), file_size, encryption_time])

# Create DataFrame
df = pd.DataFrame(
    results,
    columns=["File", "File Size (bytes)", "Encryption Time"]
)

# Display table
print(df)

# Create graph
plt.figure(figsize=(8,5))

plt.plot(
    df["File Size (bytes)"],
    df["Encryption Time"],
    marker='o'
)

plt.title("File Size vs AES Encryption Time")
plt.xlabel("File Size (bytes)")
plt.ylabel("Encryption Time (seconds)")

plt.grid(True)

plt.show()
```
